```python
from unknot_repo import available_unknots as unknots #This file holds many examples used in our paper
from drawing_rectangular_diagrams_src import draw_rectangular_diagram as draw
from push_Ts import *
from hinge_Ts import *
```

```python
def simplify_v1b_T_and_rotate(V):
    rotations=0
    while True:
        complexity=len(V)//2
        if complexity==2:
            print('The diagram has been unknotted!')
            return V
        VS=get_v1b_reduced_diagram(V)
        V=VS[-1]
        if len(V)//2<complexity:
            rotations=0
            continue
        print('The above diagram is v1b reduced. Attempting push T moves')
        V=perform_push_T_move_and_v1b_reduce(V)
        if len(V)//2<complexity:
            rotations=0
            continue
        print('The above diagram is v1b reduced and all push T moves attempted. Attempting hinge T moves')    
        V=perform_hinge_Ts_and_simplify(V)
        if len(V)//2<complexity:
            rotations=0
            continue
        if rotations<=3:
            print('Performing a 90 degree rotation of diagram and attempting all moves again')
            V = rotate_vertex_set_90_ccw(V)
            rotations=rotations+1
            continue
        else:
            print('The diagram is still knotted.')
            return V
```
